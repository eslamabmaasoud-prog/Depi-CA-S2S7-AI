```python
def info_user(nam: str, age : int)-> str:  
    '''  
    this function will return the name and age of the user
    args:
       name(str): name of the user 
       age (int): age of the user
    Returns:
       str: name and age of the user    
    '''
    return f"my name {nam}and my age is{age}"
```

```python
print(info_user("amit", 20))
info_user(age = 20,nam = "amit")
```

```python
def train_model(model_name,*datasets,**hyperparameter):
    print(f"trainmodel: {model_name}")

    print("\ndatasets")
    for dataset in datasets:
        print("-",dataset)

    print ("\nHyperparameter")
    for key , value in hyperparameter.items():
        print(f"{key},{value}")

```

```python
train_model(
   "froud detection",
   "train.csv",
   "validation.csv",
   "test.csv",
   learning_rate=0.001,
   epochs=100,
   batch_size=64,
   optmizer="Admen",
   early_stopping=True

)
```
